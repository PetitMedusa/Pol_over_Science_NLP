```python
import pandas as pd
import re
```

```python
# Load the CSV file
df = pd.read_csv('ARG/Unculled_central_tweets/community_0_1Kmostcentral_2025-05-19.txt')

# Function to clean text
def clean_text(text):
    # Convert to string in case there are non-string values
    text = str(text)
    
    # Remove RT, @mentions and &AMP
    text = re.sub(r'RT\s+', '', text)  # Remove RT
    text = re.sub(r'@\w+', '', text)   # Remove @mentions
    text = re.sub(r'&AMP;', '', text)  # Remove &AMP;
    text = re.sub(r'&amp;', '', text)  # Remove &amp;
    
    # Remove URLs
    text = re.sub(r'https?://\S+', '', text)
    
    # Remove non-ASCII characters
    text = re.sub(r'[^\x00-\x7F]+', '', text)
    
    # Remove extra spaces
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
# Apply cleaning to each tweet
df['cleaned_tweet'] = df['tweet_text'].apply(clean_text)
```

```python
# Combine all cleaned tweets into a single text with newlines between them
combined_text = '\n'.join(df['cleaned_tweet'].tolist())
```

```python
# Write to a text file
with open('USA/Central_Tweets/USA_1kcentral_com2', 'w', encoding='utf-8') as f:
    f.write(combined_text)
    print("Cleaned and combined text file created successfully!")
```
